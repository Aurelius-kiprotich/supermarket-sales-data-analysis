```python


import matplotlib.pyplot as plt
import numpy as np

# Step 1: Create pivot table
pivot = pd.pivot_table(
    df,
    values="Total",
    index="Gender",
    columns="Product line",
    aggfunc="sum"
)

# Step 2: Plot heatmap
fig, ax = plt.subplots(figsize=(10, 6))

im = ax.imshow(pivot.values)

# Step 3: Set labels
ax.set_xticks(range(len(pivot.columns)))
ax.set_yticks(range(len(pivot.index)))

ax.set_xticklabels(pivot.columns, rotation=45, ha="right")
ax.set_yticklabels(pivot.index)

# Step 4: Add numbers inside cells
for i in range(len(pivot.index)):
    for j in range(len(pivot.columns)):
        ax.text(j, i, f"{pivot.values[i, j]:.0f}",
                ha="center", va="center", color="white")

# Step 5: Titles
ax.set_xlabel("Product Line")
ax.set_ylabel("Gender")
ax.set_title("Total Sales Heatmap: Product Line vs Gender")

plt.tight_layout()
plt.show()
```

```python
#Identify the top 5 highest-selling invoices and visualize their sales

sales = df["Total"]
invoices = df["Invoice ID"]

highest_selling = list(zip(invoices,sales))

top_5 = sorted(highest_selling,key=lambda x: x[1], reverse = True)[:5]

top_sale = []
top_invoices = []

top_invoices = [invoice for invoice, sale in top_5]
top_sales = [sale for sale, sale in top_5]

plt.bar(top_invoices,top_sales) 
plt.xlabel("Top 5 invoices")
plt.ylabel("Sales")
plt.title("Highest selling invoices")

plt.xticks(rotation = 30)

plt.show()
```

```python
#Compare average sales between Members and Normal customers using a bar chart

customer = df["Customer type"]
sale = df["Total"]

avarage_sale = list(zip(customer,sale))

member_total = []
normal_total = []

for customer, sale in avarage_sale:
    if customer == "Member":
        member_total.append(sale)
    elif customer == "Normal":
        normal_total.append(sale)


member_avarage = sum(member_total)/len(member_total)

normal_avarage = sum(normal_total)/len(normal_total)

avarage = [
    member_avarage,
    normal_avarage
]

custom = [
    "Member",
    "Normal"
]

plt.bar(custom, avarage)
plt.xlabel("Customer")
plt.ylabel("Avarage sales")
plt.title("Avarage sale per customer")

plt.show()
```
